## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
df.columns.tolist()
```

```python
import pandas as pd
import numpy as np

# 1. Load starter data from data/raw/
df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

# 2. Define Unit of Analysis (1 Row = 1 Piece of Content / URL)
unit_of_analysis = df[[
    'content_id', 
    'content_type', 
    'search_volume', 
    'impressions_90d', 
    'trend_direction'
]].copy()

# 3. Calculate Proxy Target (Opportunity Score)
# Prioritize content with high demand that is experiencing a downward trend
unit_of_analysis['opportunity_score'] = np.where(
    unit_of_analysis['trend_direction'] == 'down',
    unit_of_analysis['impressions_90d'] * np.log1p(unit_of_analysis['search_volume']),
    0.0
)

# 4. Sort by top opportunity priority
unit_of_analysis = unit_of_analysis.sort_values(by='opportunity_score', ascending=False)

# 5. Display top 5 rows to prove unit of analysis and target format
unit_of_analysis.head()
```


+## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Why Fixed Rules Fail:
A naive rule like "update all pages older than 180 days" fails because it treats a zero-traffic page the same as a high-value landing page. Conversely, filtering solely by high impressions ignores decay rates and seasonality.

Why ML/Scoring Wins:
ML captures non-linear interactions between multiple dynamic signals (CTR drift, search volume decay, seasonality, and update age). It adapts dynamically to traffic patterns rather than relying on arbitrary static thresholds, maximizing editorial efficiency.
